## Implementation Notes

Production quant code usually fails in quiet places: the wrong measure, discounting at the wrong time, using future information in an exercise decision, or reporting a Monte Carlo price without an error bar. As you change the inputs, keep asking which probability measure the expectation belongs to, which information set is available at the decision time, and what limiting case should be used as a benchmark.

For interview preparation, do not stop at a number. Explain the algorithm, its complexity, and the financial reason the implementation is valid.
