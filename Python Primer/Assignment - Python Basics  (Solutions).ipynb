## 3. Check Sorting

Write a function that takes a list of numbers as input and prints 'Ascending' if the list is sorted in ascending order, 'Descending' if the list is sorted in descending order , 'Not sorted' if the list is not sorted in any manner

```python
def sortCheck(x):
    
    x_ascending = x.copy()
    x_descending = x.copy()
    
    x_ascending.sort()
    x_descending.sort(reverse = True)
    
    if x == x_ascending:
        print("Ascending")
        
    elif x == x_descending:
        print("Descending")
        
    else:
        print("Not sorted")
        
sortCheck([1,2,3,10])
sortCheck([10,8,3,-1])
sortCheck([2,3,-1,5])
```

## 4. Greatest Divisor

Write a function that takes a number as input and returns the greatest divisor of the number different from itself. e.g. f(21) = 7

```python
def f(x):
    greatestDivisor = x - 1
    while True:
        if x % greatestDivisor == 0:
            return greatestDivisor
            break
        else:
            greatestDivisor -= 1
            
f(24)
```

## 5. Shortest Word in a Sentence

Write a function that takes a sentence as input and prints the length of the shortest word in that sentence. e.g. f('Honesty is the best policy') = 2

```python
def f(x):
    words = x.split(" ")      # splitting by space
    minLength = 9999          # Initializing a large minLength value
    
    for word in words:
        if len(word) < minLength:
            minLength = len(word)   # if a word is shorter than the current minLength, it's length becomes the new minLength
    print(minLength)
    
f('Honesty is the best policy')
f('Hello World')
```

## 16. Leap Year

Write a function that returns the total number of leap years between and including a start year and an end year. The function will accept these two years as arguments. e.g. f (2003, 2008) = 2 as 2004 and 2008 are leap years

```python
def leapYearCounter(start_year, end_year):
    counter = 0
    for x in range(start_year, end_year+1):
        if x % 100 == 0:
            if x % 400 == 0:
                counter += 1
        elif x % 4 == 0:
            counter += 1
    return counter

leapYearCounter(2003, 2008)
```

```python
def fibonacci(n):
    if n == 1:
        return 0
    elif n == 2:
        return 1
    else:
        return fibonacci(n-1)+fibonacci(n-2)

print(fibonacci(30))

# golden ratio is 1.61803399

x = fibonacci(30)
y = fibonacci(31)

print(y/x)
```

## 27. The Line Class

Create a Class called line which needs two properties for initialization (m , c). Create two methods called *x_intercept()* and *y_intercept()*. When these methods are called, they should return *x_intercept value* and *y_intercept value* of the line respectively. Create another method called *Point_lies(x,y)* that accepts the cordinates of a point and returns True if the point lies on the line and False otherwise.

```python
class line:
    def __init__(self,m,c):
        self.m = m
        self.c = c
        
    def x_intercept(self):
        return (-self.c / self.m)
    
    def y_intercept(self):
        return self.c
    
    def Point_lies(self,x,y):
        if y == (self.m)*x + self.c:
            return True
        else:
            return False

l = line(2,3)
print(l.x_intercept())
print(l.y_intercept())
print(l.Point_lies(3,9))
```

# 29. Max Product

You have infinite supply of 1's,  2's and 3's. Write a function that takes a number > 10 as input and breaks it down to 1's 2's and 3's such that the product is maximum. The function returns the max product e.g. f(13) = 108 which is 3 x 3 x 3 x 2 x 2

```python
# It can be solved analytically
# x^(1/x) attains maxima at x = e = 2.718. The closest integer to e is 3
# A fixed sum s can be broken down to (s/n) number of n's
# the product will be n^(s/n) = [n^(1/n)]^s 
# since s is fixed, it boils down to n^(1/n) and we know it will be maximum if n = 3
# So a fixed sum must be broken down to as many 3's as possible to maximize the product
# The only exception is 4 should be kept as 4 because 3 x 1 < 4

def max_product1(x):    # using recursion
    if x == 1:
        return 1
    elif x == 2:
        return 2
    elif x ==3:
        return 3
    elif x ==4:
        return 4
    else:
        return 3*max_product1(x-3)
    
max_product1(13)

# The above logic can be summarized as 
# if x is a multiple of 3, then extract all 3's
# if x is of the form 3k + 1, then extract (k-1) 3's and leave one 4
# if x is of the form 3k + 2, then extract k 3's and a 2

def max_product2(x):
    if x == 1:
        return 1
    elif x == 2:
        return 2
    elif x % 3 == 0:
        return 3**(x/3)
    elif x % 3 == 1:
        return 4*(3**((x-1)/3-1))
    else:
        return 2*(3**((x-2)/3))

max_product2(13)
```

## 30. Gradient Descent

Use the algorithm of gradient descent used in class to solve a minimization problem. Write a function that accepts six arguments (a,b,c, eta, runs, guess_x) where the coefficients (a,b,c) have the usual meanings in a quadratic polynomial. eta stands for learning rate. runs stand for number of iterations the function will perform. guess_x is the initial seed value for x. The function will iterate over starting with the guess value and print the minima after specified number of runs. (a is given as positive always)

```python
def grad_descent(a,b,c,eta,runs,guess_x):
    x = guess_x
    Iterations = 0
    while(Iterations < runs):
        gradient = 2*a*x + b             # this is the slope of ax^2 + bx + c
        x -= gradient*eta
        Iterations += 1
    print("solution x : %f and function minima %f" %(x, a*(x**2)+b*x+c))
    
grad_descent(1,-6,9,0.2,100,2)
```
